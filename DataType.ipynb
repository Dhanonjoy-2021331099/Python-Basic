```python
# Built in DataType
''' 
int = 10,5
float = 3.1416, -7.8
str = "String", "Name" 
bool = True, False
list = [1,2,5,4]
tuple = (1,2,4,5)
dict = {"name":"Jibon"}
set = {1,2,3,4} # unique value
NoneType = None 

'''
```

```python
x = 5
print(type(x)) # <class 'int'>

y = "Hello"
print(type(y)) # <class 'str'>

pi = 3.1416
print(type(pi)) # <class 'float'>

is_alive = True
print(type(is_alive)) # <class 'bool'>

fruits = ["apple","Banana","Cherry"]
print(type(fruits)) # <class 'list'>

point = (10,20)
print(type(point)) # <class 'tuple'>

student = {"name": "Dhanonjoy", "age": 22}
print(student["name"])         # Dhanonjoy
print(type(student))           # <class 'dict'>

nums = {1, 2, 3, 3, 2}
print(nums)           # Output: {1, 2, 3}
print(type(nums))     # <class 'set'>

x = None
print(x)
print(type(x))  # <class 'NoneType'>
```

```python
# range type // 0 to (n-1)
# range(start, stop, step)
x = range(6) # 0 to 5

print(x)
print(list(x))
print(set(x)) # mutable
print(tuple(x))
# print(dict(x)) # Error

r = range(2, 10, 2)
print(list(r))
# range(start, stop, step) → 2 থেকে 10 এর আগ পর্যন্ত, 2 করে বাড়বে
# So, 3,4,6,8
```

```python
# frozenset (immutable-> not changable, add or remove kora jabena)
x = frozenset({"apple","Banana","Cherry"})
print(x)
# x.add("mango") # immutable
```

```python
# bytes =>For  Binary data handle (immutable)
x = bytes(10)
# x[4] = 1 # Error, immutable issue
print(x)
```

```python
# bytearray => mutable
x = bytearray(5)
print(x)
x[3] = 50 # index a value assign
print(x)
```

```python
# memoryview => Binary data k eksthe access korte dey
x = memoryview(bytes(5))
print(x)
print(x[4])

# memoryview কাজ করে বড় বড় বাইনারি ডেটার efficient access এর জন্য, যেমন — image বা file buffer।
```
