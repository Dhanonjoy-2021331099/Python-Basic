```python
# String = Collection of Characters
x = "Hello"
y = "world"
print(x,y)
```

```python
# Multiline String

msg = """ Hello, everyone 
how are you?
"""
print(msg)
```

```python
# String as Array indexing

txt= "Python"
print(txt[0])
print(txt[2])
print(txt[5])
```

```python
# Loop through String (String interable)
name = "Dhanonjoy"
for i in name:
    print(i)
```

```python
# String length
txt = "ChatGPT"
length = len(txt)
print(length)
```

```python
# subsrting exists (boolean)
txt = "I Love Coding"

sub = "Love" in txt
print(sub)

sub = "   " in txt
print(sub)

sub = " " in txt
print(sub) 
  
sub = "You" in txt
print(sub)
```

```python
# String Slicing

txt = "Artificial"

s1 = txt[ :3]
print(s1) # Art

s2 = txt[3:7]
print(s2) # ific

s3 = txt[ 3:3]
print(s3) # null

s4 = txt[4:len(txt)]
print(s4)
```

```python
# Modify String (Upper, Lower,Strip, Replace,Split)
a = "Machine learning"

s1 = a.upper()
print(s1)
s2 = a.lower()
print(s2)
s3 = a.strip()
print(s3)

s4 = a.replace('M','H')
print(s4)

s5 = a.split(" ")
print(s5)
```

```python
# String Concatenation
a = "Machine"
b = "Learning"
s = a +" "+ b
print(s)
```

```python
# String Formating Variable (f'string formate process)
name = "Machine Learning"
algo = 50
txt = f"Learn {name}, algorithm {algo}"
print(txt)
```

```python
# capitalize() # first character capital 
a = "machine learning"
s1 = a.capitalize()
print(s1)
```

```python
# count() # character number count korbe
a = "machine learning"
x = a.count('i')
print(x)
```

```python
# endswith("txt") # ses string konta # True /False
filename = "jibon.txt"
print(filename.endswith('txt'))
```

```python
# startswith("Py")
# স্ট্রিং এই দিয়ে শুরু হয় কিনা?
txt = "Python is powerful"
print(txt.startswith("Py"))   # True
print(txt.startswith("py"))   # False (case sensitive)
```

```python
# find("on") # substring koto index theke suru hyec
a = "machine learning"
x = a.find("no")
print(x)
x = a.find("ne")
print(x)
```

```python
# isalnum() 
# শুধু অক্ষর ও সংখ্যা আছে কিনা (special char বাদে)।
# return True / False

a = "Python3"
b = "Python !!!"
c = "Python 3 " # space is a special character
print(a.isalnum())
print(b.isalnum())
print(c.isalnum())
```

```python
# isalpha()
# স্ট্রিং-এ শুধু অক্ষর আছে কিনা? Return True/ False
a = "Python3"
b = "Python"
c = "Python 3 " # space is a special character
print(a.isalpha())
print(b.isalpha())
print(c.isalpha())
```

```python
#isdigit()
# স্ট্রিং-এ শুধু সংখ্যা?
txt1 = "12345"
txt2 = "123abc"
print(txt1.isdigit())  # True
print(txt2.isdigit())  # False
```

```python
# islower()
# সব অক্ষর ছোট হাতের কিনা?
txt1 = "hello"
txt2 = "Hello"
print(txt1.islower())  # True
print(txt2.islower())  # False
```

```python
#isupper()
#  সব অক্ষর বড় হাতের কিনা?
txt1 = "HELLO"
txt2 = "Hello"
print(txt1.isupper())  # True
print(txt2.isupper())  # False
```
