```python
# Changing DataType
#  Type Conversion (Casting)
x = 5
print(type(x)) # <class 'int'>

x = str(x)
print(type(x)) #<class 'str'>

x = int(x)
print(type(x)) # <class 'int'>

x = float(x)
print(type(x)) # <class 'float'>
```

```python
# Variable Scope

# Global Variable (Declare : Outside the function, access from any position)
x = 100
def show():
    print("Inside function : ",x) # accessable
show()
print("Outside fuction : ",x) # accessable


# Local Variable (Declate inside the function, just accessable own function)
def local():
    yy = 500
    print("Inside function : ",yy) # accessable
local()
# print("Outside Function : ",yy) # Error! yy is not defined

```
